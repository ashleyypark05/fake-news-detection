## 3. Step-by-Step Preprocessing

We apply three cleaning steps in sequence:

1. **Punctuation removal** — strip Unicode curly quotes/dashes and ASCII punctuation
2. **Stopword removal** — drop NLTK English stopwords
3. **Lemmatisation** — reduce each token to its dictionary base form

All logic lives in `src/preprocessing.py`; this notebook just orchestrates it.

### 3.1 Punctuation Removal

```python
df_clean = df[['title', 'text', 'label']].copy()

df_clean['title'] = df_clean['title'].apply(remove_punctuation)
df_clean['text']  = df_clean['text'].apply(remove_punctuation)

print('Before:', df['text'].iloc[4][:120])
print()
print('After: ', df_clean['text'].iloc[4][:120])
```

### 3.2 Stopword Removal

```python
df_clean['title'] = df_clean['title'].apply(remove_stopwords)
df_clean['text']  = df_clean['text'].apply(remove_stopwords)

print('After stopword removal:', df_clean['text'].iloc[4][:120])
```

### 3.3 Lemmatisation

```python
df_clean['title'] = df_clean['title'].apply(lemmatize)
df_clean['text']  = df_clean['text'].apply(lemmatize)

print('After lemmatisation:', df_clean['text'].iloc[4][:120])
```

## 4. Quality Check

```python
null_counts = df_clean[['title', 'text']].isnull().sum()
print('Null counts after cleaning:')
print(null_counts)
print(f'\nRows retained: {len(df_clean)} / {len(df)}')
df_clean.head()
```

## 5. Save

```python
df_clean.to_csv('../data/lemmatized.csv', index=False)
print('Saved to data/lemmatized.csv')
```
